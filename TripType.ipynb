## Import Libraries

```python
import pandas as pd
import numpy as np 
```

## Load Data

```python
train_data = pd.read_csv('train_data.csv')
train_data.head()
```

```python
test_data = pd.read_csv('test_data.csv')
```

## Preprocessing

```python
#1-delete some attributes
train_data.drop(['Created','CancelTime','BillID','TicketID','UserID','HashPassportNumber_p', 'HashEmail',
                  'BuyerMobile','NationalCode'], axis=1, inplace=True)
test_data.drop(['Created','CancelTime','BillID','TicketID','UserID','HashPassportNumber_p', 'HashEmail',
                  'BuyerMobile','NationalCode'], axis=1, inplace=True)
```

```python
#2-checking null values
train_data.isna().sum()
test_data.isna().sum()

#3-fill null values
train_data['VehicleClass'].fillna(train_data['VehicleClass'].mode()[0], inplace=True)
test_data['VehicleClass'].fillna(train_data['VehicleClass'].mode()[0], inplace=True)
#other columns still remain
```

```python
#4-add some attributes and encode

train_data['Male'] = train_data['Male'].astype(int)
test_data['Male'] = test_data['Male'].astype(int)

train_data['VehicleClass'] = train_data['VehicleClass'].astype(int)
test_data['VehicleClass'] = test_data['VehicleClass'].astype(int)


train2 = pd.get_dummies(train_data['Vehicle'])
train_data = pd.concat([train_data, train2], axis=1)
train_data.drop(['Vehicle'],axis=1,inplace=True)

test2 = pd.get_dummies(test_data['Vehicle'])
test_data = pd.concat([test_data, test2], axis=1)
test_data.drop(['Vehicle'],axis=1,inplace=True)


# train_data
```

```python
train_data.drop(['From','To','Domestic'],axis=1,inplace=True)
test_data.drop(['From','To','Domestic'],axis=1,inplace=True)
#delete Domestic cause InternationalPlane is exactly vice versa. so one of them should be delete.
# train_data
```

```python
train_data['FinalPrice'] = train_data['Price']-train_data['CouponDiscount']
test_data['FinalPrice'] = test_data['Price']-test_data['CouponDiscount']

train_data.drop(['Price','CouponDiscount'],axis=1,inplace=True)
test_data.drop(['Price','CouponDiscount'],axis=1,inplace=True)

train_data.head()
```

```python
train_data
```

```python
from datetime import date

train_data['DepartureTime'] = pd.to_datetime(train_data['DepartureTime'])
train_data['Day'] = train_data['DepartureTime'].dt.day 
train_data['Month'] = train_data['DepartureTime'].dt.month 
train_data['S1'] = train_data['Month']*100
train_data['S2'] = train_data['S1']+train_data['Day']

conditions = [ (train_data['S2'] >= 301) & (train_data['S2'] <= 531),
               (train_data['S2'] > 531) & (train_data['S2'] < 901),
               (train_data['S2'] >= 901) & (train_data['S2'] <= 1130),
               (train_data['S2'] > 1130) , (train_data['S2'] <= 229) ]
values = ['Spring', 'Summer', 'Fall', 'Winter', 'Winter']

train_data['Season'] = np.select(conditions, values)
train_data.drop(['DepartureTime','Day','Month','S1','S2'],axis=1,inplace=True)



test_data['DepartureTime'] = pd.to_datetime(test_data['DepartureTime'])
test_data['Day'] = test_data['DepartureTime'].dt.day 
test_data['Month'] = test_data['DepartureTime'].dt.month 
test_data['S1'] = test_data['Month']*100
test_data['S2'] = test_data['S1']+test_data['Day']

conditions = [ (test_data['S2'] >= 301) & (test_data['S2'] <= 531),
               (test_data['S2'] > 531) & (test_data['S2'] < 901),
               (test_data['S2'] >= 901) & (test_data['S2'] <= 1130),
               (test_data['S2'] > 1130) , (test_data['S2'] <= 229) ]
values = ['Spring', 'Summer', 'Fall', 'Winter', 'Winter']

test_data['Season'] = np.select(conditions, values)
test_data.drop(['DepartureTime','Day','Month','S1','S2'],axis=1,inplace=True)

train_data.head()
```

```python
train_data['Season'].value_counts()
```

```python
train3 = pd.get_dummies(train_data['Season'])
train_data = pd.concat([train_data, train3], axis=1)
train_data.drop(['Season'],axis=1,inplace=True)

test3 = pd.get_dummies(test_data['Season'])
test_data = pd.concat([test_data, test3], axis=1)
test_data.drop(['Season'],axis=1,inplace=True)
```

```python
train_data.drop(['VehicleType'],axis=1,inplace=True)
test_data.drop(['VehicleType'],axis=1,inplace=True)
```

```python
train_data
```

## Machine Learning Model

```python
# modeling 

from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
model = AdaBoostClassifier(DecisionTreeClassifier(max_depth=5), n_estimators=350)
df = train_data.drop(['TripReason'],axis=1)
model.fit(df, train_data['TripReason'])
```

## Evaluation

```python
# evaluate model
from sklearn.metrics import  f1_score

y_pred = model.predict(df)
f1_score(train_data['TripReason'], y_pred, average='binary', pos_label='Work')
```

## Prediction

```python
# predict test samples

class_pred = model.predict(test_data)
submission = pd.DataFrame(class_pred)
submission.columns = ['TripReason']
submission
```
